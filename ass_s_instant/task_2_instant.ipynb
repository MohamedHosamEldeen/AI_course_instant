```python

setA = {10, 20, 30, 40, 50}
setB = {30, 40, 60, 70}

common = setA.intersection(setB)
print("Common elements:", common)

only_in_A = setA.difference(setB)
print("Only in setA:", only_in_A)

combined = setA.union(setB)
print("Union:", combined)

setA.add(90)
setA.remove(10)
print("Updated setA:", setA)
```
