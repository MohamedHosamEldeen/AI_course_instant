# Check Prime Number
* ## Ask the user for a number and check if it is a prime number using a loop.

```python
n = int(input("Enter the number: "))
if n <= 1:
    print("Not a Prime Number")
else:
    for i in range(2, n):
        if n % i == 0:
            print("Not a Prime Number")
    else:
        print("Prime Number")
```

# Count Vowels in a String
## Take a string input and count how many vowels it contains.

```python
count = 0
word = input("enter your word: ")
for i in range(len(word)):
    if word[i] == "a" or word[i] == "e" or word[i] == "i" or word[i] == "o" or word[i] == "u":
        count = count + 1
print("Number of vowels:", count)

```

# Vowel or Consonant using Switch

## Take a single alphabet letter.

* ## If it’s "a" | "e" | "i" | "o" | "u" → print "Vowel"

* ## Else → print "Consonant"

```python

ch = input("Enter a single alphabet letter: ").lower()
match ch:
    case "a" | "e" | "i" | "o" | "u":
        print("Vowel")
    case _:
        print("Consonant")
```

# Rock-Paper-Scissors

## Write a program for rock-paper-scissors game using match-case.

* ### rock vs scissors → rock wins

* ### scissors vs paper → scissors wins

* ### paper vs rock → paper wins

* ### same choice → draw

```python
player1 = input("Player 1, enter your choice (rock/paper/scissors): ").lower()
player2 = input("Player 2, enter your choice (rock/paper/scissors): ").lower()
valid_choices = ["rock", "paper", "scissors"]

if player1 not in valid_choices or player2 not in valid_choices:
    print("Invalid input! Please choose rock, paper, or scissors.")
else:
    match (player1, player2):
        case (a, b) if a == b:
            print("Draw")
        case ("rock", "scissors") | ("scissors", "rock"):
            print("Rock wins")
        case ("scissors", "paper") | ("paper", "scissors"):
            print("Scissors wins")
        case ("paper", "rock") | ("rock", "paper"):
            print("Paper wins")
```
