# Problem 1: Student Grades Analyzer

## Write a function analyze_grades(*grades) that:

* ## Calculates the average grade.

* ## Uses match-case to return:

* * ## "Excellent" if avg ≥ 90

* * ## "Good" if avg ≥ 75

* * ## "Pass" if avg ≥ 50

* * ## "Fail" otherwise
  * analyze_grades(90, 80, 100) → "Excellent"


```python
def analyze_grades(*grades):
    avg = sum(grades) / len(grades)
    match avg:
        case x if x >= 90:
            return "Excellent"
        case x if x >= 75:
            return "Good"
        case x if x >= 50:
            return "Pass"
        case _:
            return "Fail"
```

# Problem 2: Flexible Calculator

## Write a function calculator(operation, *args, precision=2) that:

* ## Supports "sum", "multiply", "max", "min", "average".

* ## Uses a loop (for) to compute results.

* ## Uses match-case to decide the operation.

* ## Round to precision decimals.

* calculator("average", 10, 20, 30) → 20.0


```python
def calculator(operation, *args, precision=2):
    if operation == "sum":
        result = sum(args)
    elif operation == "multiply":
        result = 1
        for n in args:
            result *= n
    elif operation == "max":
        result = max(args)
    elif operation == "min":
        result = min(args)
    elif operation == "average":
        result = sum(args) / len(args)
    else:
        return "Invalid operation"

    return round(result, precision)
```

* ## Write a function generate_primes(n) that uses a while loop to return all prime numbers up to n.
* ## 👉 Then use a lambda with filter() to keep only primes that are odd.
* generate_primes(10) → [2, 3, 5, 7]
* odd primes → [3, 5, 7]


```python
def generate_primes(n):
    primes = []
    num = 2
    while num <= n:
        is_prime = True
        for i in range(2, num):
            if num % i == 0:
                is_prime = False
                break
        if is_prime:
            primes.append(num)
        num += 1
    return primes

odd_primes = lambda nums: list(filter(lambda x: x % 2 != 0, nums))
```

# Problem 5: Employee Salary Manager

## Write a function employee_salary(base, *bonuses, tax=0.1, **deductions) that:

* ## Adds all bonuses using a loop.

* ## Subtracts all deductions (sum of kwargs values).

* ## Applies tax on final salary.

* ## Uses conditions: if salary < 3000 → "Low", else "High".

* employee_salary(5000, 200, 300, tax=0.15, insurance=500, loan=200)


```python
def employee_salary(base, *bonuses, tax=0.1, **deductions):
    salary = base + sum(bonuses) - sum(deductions.values())
    salary -= salary * tax
    status = "Low" if salary < 3000 else "High"
    return salary, status
```

# Problem 6: Math Game with Switch

## Write a function math_game(choice, *numbers) that:

## Uses match-case where:

* ## 1: return sum of even numbers

* ## 2: return sum of odd numbers

* ## 3: return product of numbers divisible by 3

* ## _: return "Invalid choice"

* ## Use loops + conditions to calculate.

math_game(1, 1, 2, 3, 4, 5, 6) → 12   (2+4+6)


```python
def math_game(choice, *numbers):
    if choice == 1:
        return sum(n for n in numbers if n % 2 == 0)
    elif choice == 2:
        return sum(n for n in numbers if n % 2 != 0)
    elif choice == 3:
        result = 1
        for n in numbers:
            if n % 3 == 0:
                result *= n
        return result
    else:
        return "Invalid choice"
```

# Problem 7: Student Report with Switch + Loop

## Write a function student_report(**students) where keys are student names and values are grades.

* ## Use a loop to go through each student.

* ## Use match-case to print their status:

* * ## ≥ 90 → "Excellent"

* * ## ≥ 75 → "Good"

* * ## ≥ 50 → "Pass"

* * ## Else → "Fail"
* student_report(Ali=95, Omar=60, Sara=40)


```python
def student_report(**students):
    for name, grade in students.items():
        if grade >= 90:
            status = "Excellent"
        elif grade >= 75:
            status = "Good"
        elif grade >= 50:
            status = "Pass"
        else:
            status = "Fail"
        print(name, ":", status)
```

# Problem 8: Custom Filter with Lambda + Switch

## Write a function custom_filter(choice, numbers) where:

* ## choice=1: keep even numbers

* ## choice=2: keep odd numbers

* ## choice=3: keep multiples of 5

* ## Use filter() + lambda + match-case.

* custom_filter(2, [1,2,3,4,5,6,7]) → [1,3,5,7]


```python
def custom_filter(choice, numbers):
    if choice == 1:
        return list(filter(lambda x: x % 2 == 0, numbers))
    elif choice == 2:
        return list(filter(lambda x: x % 2 != 0, numbers))
    elif choice == 3:
        return list(filter(lambda x: x % 5 == 0, numbers))
    else:
        return []
```
