```python
import numpy as np
arr = np.arange(1, 13)

print(arr)

print(arr.reshape(3, 4))

print(arr.reshape(2, 6))

print("\nInvalid reshape ")
print(arr.reshape(5, 3))  
```

```python
import numpy as np

uniform_arr = np.random.random((3, 3))
normal_arr = np.random.normal(0, 100, (3, 3))

print("Uniform Random Array:\n", uniform_arr)
print("\nNormal Distribution Array:\n", normal_arr)

print(uniform_arr.shape)
print(uniform_arr.ndim)
print(uniform_arr.size)
print(uniform_arr.dtype)
```
