```python
num = int(input("Enter a number: "))

if num % 2 == 0 and num % 3 == 0:
    print("Div ")
elif num % 2 == 0:
    print("Div by 2 ")
elif num % 3 == 0:
    print("Div by 3 ")
else:
    print("Not div by 2 or 3")
```

```python
nums = [3, 9, 2]
max_num = 0

if nums[0] > max_num:
    max_num = nums[0]
    
if nums[1] > max_num:
    max_num = nums[1]
    
if nums[2] > max_num:
    max_num = nums[2]
    
print("Max num:", max_num)
```

```python
text = input("Enter a string: ")
match text:
    case a if text[0] == text[-1]:
        print("First and last characters are equal.")
    case b if text[0] != text[-1]:
        print("First and last characters are NOT equal.")
```

```python
nums = [3, 9, 2]

match nums:
    case [a, b, c] if a >= b and a >= c:
        max_num = a
    case [a, b, c] if b >= a and b >= c:
        max_num = b
    case [a, b, c]:
        max_num = c

print("Max num :", max_num)
```
