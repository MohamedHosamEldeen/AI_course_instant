```python
#Q2
def word(s):
    if len(s)  < 2:
        return ""
    return s[:2] + s[-2:]

print(word("w3resource")) 
print(word("w3"))    
print(word("w"))
```

```python
#Q3
def word(s):
    if len(s) < 3:
        return s
    elif s.endswith("ing"):
        return s + "ly"
    else:
        return s + "ing"
print(word("abc"))    
        
```

```python
#Q5
def swap(s):
    if len(s) < 2:
        return s
    return s[-1] + s[1:-1] + s[0]
print(swap('abcd'))
```

```python
def repe(lst, k):
    res =[]
    for i in lst:
        if lst.count(i) >= k and i not in res:
            res.append(i)
    return res


print(repe([4,6,4,3,3,4,3,4,8], 3))
```
